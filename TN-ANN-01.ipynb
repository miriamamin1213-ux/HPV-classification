```python
import pandas as pd

df = pd.read_csv("HPV2026.csv")

print(df.shape)

print(df.columns.tolist())

df.head()
```

```python
df.info()
```

```python
print(df["HPV Status"].value_counts())
```

```python
print(df["HPV Status"].isna().sum())
```

```python
df_train = df.dropna(subset=["HPV Status"])

print(df_train.shape)
```

```python
print(df_train.isnull().sum())
```

```python
df_tn = df.dropna(
    subset=[
        "T-stage",
        "N-stage"
    ]
).copy()

print(df_tn.shape)
```

```python
print(df_tn["T-stage"].value_counts())
```

```python
print(df_tn["N-stage"].value_counts())
```

```python
print(df_tn.isnull().sum())
```

```python
df_tn["T-stage"] = df_tn["T-stage"].replace({
    "T1":0,
    "T2":1,
    "T3":2,
    "T4":3
})
```

```python
df_tn["N-stage"] = df_tn["N-stage"].replace({
    "N0":0,
    "N1":1,
    "N2":2,
    "N3":3
})
```

```python
print(df_tn["T-stage"].value_counts())
```

```python
df_tn = df_tn[df_tn["T-stage"] != "T0"]
```

```python
print(df_tn["T-stage"].value_counts())
```

```python
print(df_tn["T-stage"].dtype)
print(df_tn["N-stage"].dtype)
```

```python
df_tn["T-stage"] = df_tn["T-stage"].astype(int)
```

```python
print(df_tn["T-stage"].dtype)
print(df_tn["N-stage"].dtype)
```

```python
print(df_tn.isnull().sum())
```

```python
tobacco_mode = df_tn["Tobacco Consumption"].mode()[0]

df_tn["Tobacco Consumption"] = (
    df_tn["Tobacco Consumption"]
    .fillna(tobacco_mode)
)
```

```python
alcohol_mode = df_tn["Alcohol Consumption"].mode()[0]

df_tn["Alcohol Consumption"] = (
    df_tn["Alcohol Consumption"]
    .fillna(alcohol_mode)
)
```

```python
performance_mode = df_tn["Performance Status"].mode()[0]

df_tn["Performance Status"] = (
    df_tn["Performance Status"]
    .fillna(performance_mode)
)
```

```python
treatment_mode = df_tn["Treatment"].mode()[0]

df_tn["Treatment"] = (
    df_tn["Treatment"]
    .fillna(treatment_mode)
)
```

```python
print(df_tn.isnull().sum())
```

```python
X = df_tn[
    [
        "Age",
        "Gender",
        "Tobacco Consumption",
        "Alcohol Consumption",
        "Performance Status",
        "Treatment"
    ]
]
```

```python
y_t = df_tn["T-stage"]

y_n = df_tn["N-stage"]
```

```python
print(X.shape)

print(y_t.shape)

print(y_n.shape)
```

```python
SEED = 44
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_t_train, y_t_test, y_n_train, y_n_test = train_test_split(
    X,
    y_t,
    y_n,
    test_size=0.20,
    random_state=SEED
)
```

```python
print(X_train.shape)

print(X_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)

X_test = scaler.transform(X_test)
```

```python
X_train_final, X_val, \
y_t_train_final, y_t_val, \
y_n_train_final, y_n_val = train_test_split(
    X_train,
    y_t_train,
    y_n_train,
    test_size=0.20,
    random_state=SEED
)
```

```python
print(X_train_final.shape)

print(X_val.shape)

print(X_test.shape)
```

```python
import torch

X_train_final = torch.FloatTensor(X_train_final)

X_val = torch.FloatTensor(X_val)

X_test = torch.FloatTensor(X_test)

y_t_train_final = torch.LongTensor(
    y_t_train_final.to_numpy()
)

y_t_val = torch.LongTensor(
    y_t_val.to_numpy()
)

y_t_test = torch.LongTensor(
    y_t_test.to_numpy()
)

y_n_train_final = torch.LongTensor(
    y_n_train_final.to_numpy()
)

y_n_val = torch.LongTensor(
    y_n_val.to_numpy()
)

y_n_test = torch.LongTensor(
    y_n_test.to_numpy()
)
```

```python
print(X_train_final.shape)

print(X_val.shape)

print(X_test.shape)

print(y_t_train_final.shape)

print(y_n_train_final.shape)
```

```python
import torch.nn as nn

class TNNet(nn.Module):

    def __init__(self):

        super().__init__()

        self.fc1 = nn.Linear(6, 64)

        self.fc2 = nn.Linear(64, 32)

        self.fc3 = nn.Linear(32, 16)

        self.relu = nn.ReLU()

        self.t_head = nn.Linear(16, 4)

        self.n_head = nn.Linear(16, 4)

    def forward(self, x):

        x = self.relu(self.fc1(x))

        x = self.relu(self.fc2(x))

        x = self.relu(self.fc3(x))

        t_out = self.t_head(x)

        n_out = self.n_head(x)

        return t_out, n_out
```

```python
model = TNNet()

print(model)
```

```python
criterion_t = nn.CrossEntropyLoss()

criterion_n = nn.CrossEntropyLoss()

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.001
)
```

```python
print(model)

print(criterion_t)

print(criterion_n)
```

```python
criterion_t = nn.CrossEntropyLoss()

criterion_n = nn.CrossEntropyLoss()

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.001
)
```

```python
print(criterion_t)

print(criterion_n)

print(optimizer)
```

```python
epochs = 500

best_val_loss = float('inf')

best_epoch = 0

for epoch in range(epochs):

    model.train()

    pred_t, pred_n = model(X_train_final)

    loss_t = criterion_t(
        pred_t,
        y_t_train_final
    )

    loss_n = criterion_n(
        pred_n,
        y_n_train_final
    )

    train_loss = loss_t + loss_n

    optimizer.zero_grad()

    train_loss.backward()

    optimizer.step()

    model.eval()

    with torch.no_grad():

        val_t, val_n = model(X_val)

        val_loss_t = criterion_t(
            val_t,
            y_t_val
        )

        val_loss_n = criterion_n(
            val_n,
            y_n_val
        )

        val_loss = (
            val_loss_t +
            val_loss_n
        )

    if val_loss.item() < best_val_loss:

        best_val_loss = val_loss.item()

        best_epoch = epoch + 1

        torch.save(
            model.state_dict(),
            "best_tn_model.pth"
        )

    if (epoch + 1) % 50 == 0:

        print(
            f"Epoch {epoch+1}, "
            f"Train={train_loss.item():.4f}, "
            f"Val={val_loss.item():.4f}"
        )

print()

print("Best Validation Loss =", best_val_loss)

print("Best Epoch =", best_epoch)
```

```python
model.load_state_dict(
    torch.load("best_tn_model.pth")
)
```

```python
model.eval()

with torch.no_grad():

    pred_t, pred_n = model(X_test)

    pred_t_class = torch.argmax(
        pred_t,
        dim=1
    )

    pred_n_class = torch.argmax(
        pred_n,
        dim=1
    )
```

```python
from sklearn.metrics import classification_report

print("T-STAGE")

print(
    classification_report(
        y_t_test.numpy(),
        pred_t_class.numpy()
    )
)
```

```python
print("N-STAGE")

print(
    classification_report(
        y_n_test.numpy(),
        pred_n_class.numpy()
    )
)
```

```python
from sklearn.metrics import confusion_matrix

print("T-STAGE")

print(
    confusion_matrix(
        y_t_test.numpy(),
        pred_t_class.numpy()
    )
)
```

```python
print("N-STAGE")

print(
    confusion_matrix(
        y_n_test.numpy(),
        pred_n_class.numpy()
    )
)
```
